```python
#그룹 객체 만들기
#DataFrame객체.groupby

import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age","sex","class","fare","survived"]]
df.head()
```

```python
df["class"].unique()
```

```python
df["class"].value_counts()
```

```python
df["class"].nunique()
```

```python
grouped = df.groupby(["class"],observed=True)
grouped
```

```python
for key, group in grouped:
    print(key)
    print(group.head())
    print()
```

```python
average = grouped.mean(numeric_only=True)
average
```

```python
group3 = grouped.get_group(("Third",))
group3
```

```python
# 여러 열을 기준으로 그룹화

grouped_two = df.groupby(["class","sex"])
grouped_two
```

```python
for key, group in grouped_two:
    print(key)
    print(group.head())
    print()
```

```python
grouped_two.mean(numeric_only=True)
```

```python
# group객체.aggregate(매핑함수)
# group객체.agg(매핑함수)

df.groupby("class").aggregate("mean",numeric_only=True)
# df.groupby("class").mean()
df.groupby("class")[["fare","survived"]].aggregate("mean")
df.groupby("class")[["fare","survived"]].agg("mean")
```

```python
df.groupby("class").agg(["min","max"])
```

```python
def min_max(x):
    return x.max() - x.min()
df.groupby("class")[["age","fare"]].agg(min_max)
```
