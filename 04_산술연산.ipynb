```python
# NaN 값이 있는 시리즈 연산

import pandas as pd
import numpy as np
```

```python
student1 = pd.Series({'국어':np.nan, '영어':80, "수학":90})
student2 = pd.Series({'수학':80, "국어":90})

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
subtraction = student1 - student2
multiplication = student1 * student2
divison = student1 / student2

print(addition)
print(subtraction)
print(multiplication)
print(divison)
```

```python
restult = pd.DataFrame([
    addition,
    subtraction,
    multiplication,
    divison
], index=["덧셈","뺼셈","곱셈","나눗셈"])

restult
```
