```python
import pandas as pd

data = {
        '이름': ['김철수', '이영희', '박민수', '최지우'],
        '나이': [20, 21, 22, 20],
        '성별': ['남', '여', '남', '여'],
        '수학': [90, 85, 70, 100],
        '영어': [85, 95, 75, 90]
}


df = pd.DataFrame(data)
df
```

```python
df.head(2)
```

```python
df.info()
```

```python
df = df.set_index('이름')
```

```python
result = df.loc['이영희', ['수학', '영어']]
result
```

```python
result = df.iloc[0:2, 1:]
result
```

```python
data = pd.Series({
    '이름': ['김철수', '이영희', '박민수', '최지우'],
    '나이': [20, 21, 22, 20],
    '성별': ['남', '여', '남', '여'],
    '수학': [90, 85, 70, 100],
    '영어': [85, 95, 75, 90]
 })

sr = pd.Series(data)
print(sr.index)
print(sr.values)
```

```python
list_data = ["2010-01-02", 3.14, "ABC", 100, True]

sr = pd.Series(list_data)
sr.dtype
```

```python
tup_data = ("영인", "2010-05-01", "여", True)
index_labels = ["이름", "생년월일", "성별", "학생여부"]

sr = pd.Series(tup_data, index = index_labels)

print(sr)
```

```python
print(sr.size)
print(sr.shape)
print(sr.ndim)
```

```python
print(sr.iloc[0])
```

```python
print(sr["이름"])
print(sr.loc["이름"])
```

```python
print(sr.iloc[[1, 2]])
```

```python
print(sr[["생년월일", "성별"]])
print(sr.loc[["생년월일", "성별"]])
```
