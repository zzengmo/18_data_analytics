```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
df = titanic.loc[0:9,["age","fare"]]
df
```

```python
#불린 인덱싱

df["age"] < 20
```

```python
cond = df["age"] < 20
df[cond]
```

```python
df.loc[cond]
```

```python
cond = df["age"] < 20
df[~cond]
```

```python
#나이가 10대인 승객 을 찾습니다.

(df["age"] >= 10) & (df["age"] < 20)
```

```python
cond = (df["age"] >= 10) & (df["age"] < 20)
df[cond]
```

```python
cond = (titanic["age"] >= 10) & (titanic.age < 20)
titanic[cond]
```

```python
titanic.head(2)
(titanic["age"] < 10) & (titanic["sex"] == "female")

titanic[(titanic["age"] < 10) & (titanic["sex"] == "female")]
```

```python
#나이가 10세 미만이고 또는 60새 이상인 승객

mask3 = (titanic.age < 10) | (titanic.age >= 60)
df_under10_morethan60 = titanic.loc[mask3,["age","sex","alone"]]
df_under10_morethan60.head()
```
